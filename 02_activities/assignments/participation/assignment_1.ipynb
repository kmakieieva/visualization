```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Set random seed for reproducibility
np.random.seed(613)
   # Create array [0 - 49]
x = np.arange(50)
# Generate 50 random integers between 0 and 99
y = np.random.randint(0,100,50)

# Create figure and axes (5x3 inches)
fig, ax = plt.subplots(figsize = (5,3))
#fig, ax = plt.subplots(figsize = (5,3))
#ax.scatter(x,y)
#ax.bash(x,y)
#ax.plot(x,y)
ax.hist(y)
```

```python
# Font style for title
font1 = {'family':'sans-serif','color':'blue','size':20}
# Font style for axis labels
font2 = {'family':'monospace','color':'green','size':14}

fig, ax = plt.subplots(figsize = (5,3))
# Plot line graph
ax.plot(x,y)
# Set title with custom font
ax.set_title('Total growth over time', fontdict =font1)
# Set y axis label with custom font
ax.set_ylabel('Total growth', fontdict =font2)
# Set x axis label with custom font
ax.set_xlabel('Years since start', fontdict =font2)
fig.tight_layout()
```

```python
fig, ax = plt.subplots(figsize=(5,3))
# Create custom scatter plot
ax.scatter(x,y,marker = '*', color = 'indigo')
```

```python
x1 = np.arange(50)
y1 = np.random.randint(0,75,50)
x2 = np.array(["Luffy","Zoro","Nami","Usopp","Sanji"])
y2 = np.array([110,180,240,99,220])
```

```python
fig, (ax1, ax2) = plt.subplots(ncols = 2, nrows = 1, figsize = (7,3))
ax1.scatter(x1, y1, marker = 'o', color = 'red')
ax2.bar(x2, y2, color = 'darkgreen')
ax1.set_title('Scatterplot')
ax2.set_title('Bar Chart')
fig.show()
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1','ax3'], ['ax2','ax3']], figsize = (7,4), layout = "constrained")
someaxes['ax1'].scatter(x1,y1)
someaxes['ax2'].bar(x2,y2)
someaxes['ax3'].plot(x1,y1)

someaxes['ax1'].set_xlabel('A Big Label', fontsize = 18)
someaxes['ax2'].set_xlabel('Another Label', fontsize = 18)
someaxes['ax3'].set_xlabel('Label 2: 2 Fast 2 Furious', fontsize = 18)
```

```python
x = np.array(["Luffy","Zoro","Nami","Usopp","Sanji"])
y1 = np.array([110,180,240,99,220])
y2 = np.array([170,100,90,120,50])

fig, ax = plt.subplots(figsize=(7,3))
ax.bar(x,y1,color = 'indigo')
ax.plot(x,y2, color = 'red')
```

```python
y2_sd = np.std(y2)

fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x,y2,color = 'red')

ax.errorbar(x,y2, yerr = y2_sd, fmt = 'none', 
            ecolor = 'indigo', 
            elinewidth = 4, 
            capsize = 6, 
            capthick = 4)
```
