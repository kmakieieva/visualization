```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Set random seed for reproducibility
np.random.seed(613)
   # Create array [0 - 49]
x = np.arange(50)
# Generate 50 random integers between 0 and 99
y = np.random.randint(0,100,50)

# Create figure and axes (5x3 inches)
fig, ax = plt.subplots(figsize = (5,3))
#fig, ax = plt.subplots(figsize = (5,3))
#ax.scatter(x,y)
#ax.bash(x,y)
#ax.plot(x,y)
ax.hist(y)
```

```python
# Font style for title
font1 = {'family':'sans-serif','color':'blue','size':20}
# Font style for axis labels
font2 = {'family':'monospace','color':'green','size':14}

fig, ax = plt.subplots(figsize = (5,3))
# Plot line graph
ax.plot(x,y)
# Set title with custom font
ax.set_title('Total growth over time', fontdict =font1)
# Set y axis label with custom font
ax.set_ylabel('Total growth', fontdict =font2)
# Set x axis label with custom font
ax.set_xlabel('Years since start', fontdict =font2)
fig.tight_layout()
```

```python
x1 = np.arange(50)
y1 = np.random.randint(0,75,50)
x2 = np.array(["Luffy","Zoro","Nami","Usopp","Sanji"])
y2 = np.array([110,180,240,99,220])
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1','ax3'], ['ax2','ax3']], figsize = (7,4), layout = "constrained")
someaxes['ax1'].scatter(x1,y1)
someaxes['ax2'].bar(x2,y2)
someaxes['ax3'].plot(x1,y1)

someaxes['ax1'].set_xlabel('A Big Label', fontsize = 18)
someaxes['ax2'].set_xlabel('Another Label', fontsize = 18)
someaxes['ax3'].set_xlabel('Label 2: 2 Fast 2 Furious', fontsize = 18)
```

```python
import seaborn as sns 

np.random.seed(613)
```

```python
tips = sns.load_dataset('tips')
print(tips)
```

```python
sns.set_style('whitegrid')

sns.lineplot(data = tips, 
            x = 'total_bill', 
            y = 'tip')
```

```python
fig, ax = plt.subplots(figsize=(10, 3))

tipgraph = sns.lineplot(
    data=tips,
    x='total_bill',
    y='tip',
    color='hotpink',
    linestyle='--',
    linewidth=3,
    marker='o',
    markerfacecolor='indigo')

tipgraph.set(
    title='Tip vs Total Bill',
    xlabel='Total Bill ($)',
    ylabel='Tip Amount ($)')
```

```python
tipgraph = sns.scatterplot(data = tips,
                            x = 'total_bill',
                            y = 'tip',
                            style = 'time',
                            hue = 'day',
                            palette = ['purple','hotpink','blue','green'])

tipgraph.set(title='Tip vs Total Bill',
             xlabel='Total Bill ($)',
            ylabel='Tip Amount ($)')
```

```python
sns.pairplot(data = tips, hue = 'day')
```

```python
daysplot = sns.relplot(
    data = tips,
    x = 'total_bill',
    y = 'tip',
    hue = 'sex',
    col = 'day',
    kind = 'scatter',
    col_wrap = 2

)
```

```python
import plotly.graph_objects as go
import nbformat

x1 = np.array(["Luffy","Zoro","Nami","Usopp","Sanji"])
y1 = np.array([110,180,240,99,220])
```

```python
graph = go.Figure()
graph.add_trace(go.Bar(x=x1,y=y1))
graph.update_layout(title = 'Pirate Scores',
                    xaxis_title = 'Pirates',
                    yaxis_title = 'Score')

graph.show()
```

```python
from wordcloud import WordCloud

df = pd.read_csv('https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/movie_quotes.csv', on_bad_lines = 'skip')

df
```

```python
text = ' '.join(each for each in df.quote)

wordcloud = WordCloud(background_color = 'white',
                    colormap = 'inferno',
                    random_state = 613).generate(text)

fig, ax = plt.subplots(figsize = (7,3))
ax.imshow(wordcloud)
ax.axis('off')
```

```python
from matplotlib_venn import venn2, venn2_circles, venn2_unweighted

A = set(['apple', 'banana', 'watermelon'])
B = set(['pumpkin', 'blueberrry', 'apple', 'key lime'])

diagram = venn2([A,B],
                set_labels = ('Fruits', 'Pies'),
                set_colors = ('blue','red'),
                alpha = 0.5)

diagram.get_label_by_id('10').set_text('\n'.join(A-B))
diagram.get_label_by_id('11').set_text('\n'.join(A&B))
diagram.get_label_by_id('01').set_text('\n'.join(B-A))
```
