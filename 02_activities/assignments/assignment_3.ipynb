```python
import pandas as pd
import plotly.graph_objects as go

df = pd.read_excel("mtcg-all-population-groups-contracting-organizations-en-2024.xlsx")

print(df.head())
```

```python
print(df.columns)
```

```python
import plotly.graph_objects as go

x1 = df["Population Served"]
y1 = df["Total Operating Expenditures"]
graph = go.Figure()

graph.add_trace(go.Scatter(x=x1,y=y1,mode='markers', marker = dict(size = 10, color = 'black')))

graph.update_layout(
    title='Population Served vs Total Operating Expenditures',
    xaxis_title='Population Served',
    yaxis_title='Total Operating Expenditures',
    width=700,
    height=600
)

graph.show()
```
